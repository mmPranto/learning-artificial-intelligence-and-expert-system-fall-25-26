```python
b=20
print(a+b)
print(a-b)
print(a*b)
print(a/b)   #float division
print(a//b)  #integer division
print(a**2)  #exponent
print(2**10)
```

# if else (conditional statement)

```python
marks=90
if marks>=90:
    print("A+")
elif marks>=85 and marks<90:
    print("A")
else:
    print("I am not satisfied.")
```
