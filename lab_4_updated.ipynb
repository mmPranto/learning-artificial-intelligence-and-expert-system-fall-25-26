# Matrix(list of list)

```python
m=[
    [10,20],
    [30,40],
    [50,60]
]

print(m)
print(type(m))
print(len(m))
print(m[1])
print(len(m[0]))
```

```python
row=len(m)
col=len(m[0])
for i in range(row):
    for j in range(col):
        print(m[i][j],end=" ")
    print()
```

```python
row=4
col=5
m=[]
for i in range(row):
    r=[]
    for j in range(col):
        r.append(0)
    m.append(r)
print(m)

for r in m:
    print(r)
```

```python
def createMatrix(row,col):
    m=[]
    for i in range(row):
        r=[]
        for j in range(col):
            r.append(0)
        m.append(r)
    return m

def printMatrix(m):
    for r in m:
        print(r)

g=createMatrix(150,200)
printMatrix(g)
```

```python
def identityMatrix(val):
    m=[]
    for i in range(val):
        r=[]
        for j in range(val):
            if i==j:
                r.append(1)
            else:
                r.append(0)
        m.append(r)
    return m

k=identityMatrix(4)
printMatrix(k)
```

```python
def f(x):
    return x**2

x=[-2,-1.5,-1,-0.5,0,0.5,1,1.5,2,]
y=[]
for i in range(len(x)):
    y.append(f(x[i]))

print(x)
print(y)

import matplotlib.pyplot as plt
plt.plot(x,y)
plt.grid()
plt.show()
```
