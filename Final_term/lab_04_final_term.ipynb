# Genetic Algorithm
- Initial population
    - chromosome(Board or state in N Queen Problem)
    - Encoding(Board or state to string)
    - Find fitness

```python
import random

def createBoard(n):
    m=[]
    for i in range(n):
        r=[]
        for j in range(n):
            r.append(0)
        m.append(r)

    for c in range(n):
        r = random.randint(0,n-1)
        m[r][c]=1
    
    return m
    
def printBoard(m):
    for r in m:
        print(r)

def encoding(b):
    ch=""
    for j in range(len(b)):
        for i in range(len(b)):
            if b[i][j]==1:
                ch=ch+str(i)

    return ch
def findNumberofAttack(ch):
    attack=0
    for i in range(0,n-1):
        for j in range(i+1,n-1):
            if ch[i]==ch[j]:
                attack+=1
            if abs(int(ch[i])-int(ch[j]))==abs(i-j):
                attack+=1

    return attack

def fitness(ch):
    attack=findNumberofAttack(ch)
    max_non_attacking = n*(n-1)/2
    pair_of_non_attack=max_non_attacking-attack

    return pair_of_non_attack

n=8
b= createBoard(n)
printBoard(b)
ch=encoding(b)
print(ch)
attack=findNumberofAttack(ch)
print(attack)
non_attack=fitness(ch)
print(int(non_attack))

no_of_population =4
initial_population=[]
def population():
    for i in range(no_of_population):
        b1=createBoard(n)
        ch1=encoding(b1)
        initial_population.append(ch1)

population()
print(initial_population)
```
