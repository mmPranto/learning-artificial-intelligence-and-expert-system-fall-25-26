# Graph Representation
- Adjacency Matrix
- Adjacency List

```python
def createMatrix(n):
    m=[]
    for i in range(n):
        r=[]
        for j in range(n):
            r.append(0)
        m.append(r)
    return m

def printMatrix(m):
    for r in m:
        print(r)

f=open("graph.txt","r")
Vertex_Edge=f.readline()
v_e=Vertex_Edge.split(" ")

n=int(v_e[0])
m=int(v_e[1])
print(n,m)



g=createMatrix(n)


for i in range(m):
    line=f.readline()
    u_v=line.split(" ")
    u=int(u_v[0])
    v=int(u_v[1])
    g[u][v]=1
    #g[v][u]=1
    

f.close()
printMatrix(g)
```
