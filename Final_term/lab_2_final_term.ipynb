# Graph representation using Adjacency Matrix

```python
def createMatrix(n):
    m=[]
    for i in range(n):
        r=[]
        for j in range(n):
            r.append(0)
        m.append(r)
    return m

def printMatrix(m):
    for r in m:
        print(r)

f=open("graph_lab_2_2.txt","r")
Vertex_Edge=f.readline()
v_e=Vertex_Edge.split(" ")

n=int(v_e[0])
m=int(v_e[1])
print(n,m)

g=createMatrix(n)


for i in range(m):
    line=f.readline()
    u_v=line.split(" ")
    u=int(u_v[0])
    v=int(u_v[1])
    w=int(u_v[2])
    g[u][v]=w
    g[v][u]=w
    

f.close()
printMatrix(g)
```

# Graph representation using Adjacency List

```python
def createEmptyList(n):
    L={}
    for i in range(n):
        L[i]=[]
    return L

f=open("graph_lab_2_2.txt","r")
Vertex_Edge=f.readline()
v_e=Vertex_Edge.split(" ")

n=int(v_e[0])
m=int(v_e[1])
print(n,m)

G=createEmptyList(n)
# print(G)

for i in range(m):
    line=f.readline()
    u_v=line.split(" ")
    u=int(u_v[0])
    v=int(u_v[1])
    w=int(u_v[2])
    G[u].append((v,w))
    G[v].append((u,w))
    
    

f.close()
print(G)
```

```python
def Factorial(n):
    if n<=0:
        return 1
    else:
        return n*Factorial(n-1)

x=Factorial(5)
print(x)
    
```

```python
def GCD(n,m):
    if n==m:
        return m
    elif n>=m:
        return GCD(n-m,m)
    else:
        return GCD(n,m-n)

x=GCD(18,3)
print(x)
```
