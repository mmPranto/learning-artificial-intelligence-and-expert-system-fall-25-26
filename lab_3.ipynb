```python
List=[]
n=100
for i in range(n):
    List.append(i)
print("0 to 100 :")
print(List)

evenList=[]
oddList=[]

for i in range(n):
    if List[i]%2==0:
        evenList.append(List[i])
    else:
        oddList.append(List[i])
print("0 to 100 even numbers:")
print(evenList)
print("0 to 100 odd numbers: ")
print(oddList)
```

```python
def makedouble(li):
    dList=[]
    for i in range(len(li)):
        dList.append(li[i]*2)

    return dList;

List=[10,20,30,40,50]
result=makedouble(List)
print(result)
```

```python
def func(fl):
    fList=[]
    for i in range(len(fl)):
        cal=i**2+5*i-10
        fList.append(cal)
    return fList

List1=[-1.2,1,2.5,5,7]
res=func(List1)
print(res)
```

```python
def rightShift(li,k):
    n=len(li)
    for j in range(k):
        for i in range(n-1,0,-1):
            li[i]=li[i-1]
        li[j]=0
    return li
        

List=[10,20,30,40,50,60,70,80]
result=rightShift(List,3)
print(result)
```

```python
# left shift home work
def leftShift(li,k):
    n=len(li)
    for j in range(k):
        for i in range(0,n):
            li[i]=li[i+1]
    return li
List=[10,20,30,40,50,60,70,80]
result=leftShift(List,1)
print(result)
```

```python
# Factorial

def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)

print(factorial(5))
```

```python
# Fibonacci number


def fibo(n):
    if n==0 or n==1:
        return n
    else:
        return fibo(n-1)+fibo(n-2)


print(fibo(10))
```
